```python
red_folder = '../data/red_folder'
green_folder = '../data/green_folder'
```

```python
# Load all Csvs
chunk_size = 50_000 # manage memory

def load_all_cvs(folder_path, chunk_size, sample_frac=0.3,**kwargs):
  sampled_chunks = []
  for file in os.listdir(folder_path):
    if file.endswith('.csv'):
      file_path = os.path.join(folder_path, file)
      chunk_iter = pd.read_csv(file_path, chunksize=chunk_size, )
      for chunk in chunk_iter:
        sampled_chunk = chunk.sample(frac=sample_frac)
        sampled_chunks.append(sampled_chunk)
  return pd.concat(sampled_chunks, ignore_index=True)
```

```python
red_data = load_all_cvs(red_folder,chunk_size,sample_frac=0.2, low_memory=False)
```

```python
green_data = load_all_cvs(green_folder, chunk_size, sample_frac=0.096, low_memory=False)
```

```python
print(red_data.shape)
print(green_data.shape)
```

```python
# combine green and red datasets
full_df = pd.concat([green_data, red_data], ignore_index=True)
print("\n Shape of the full dataset:")
print(full_df.shape)
```
