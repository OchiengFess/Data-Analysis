```python
import pandas as pd
import numpy as np
import re
```

```python
#from google.colab import drive
#drive.mount('/content/drive')
```

```python
filepath = '../data/Nairobi propertyprices.csv'
df = pd.read_csv(filepath)
```

```python
df.head()
```

```python
# copy of raw
df_cleaned = df.copy()
```

###### 1. Clean 'Price' column

```python
# func

def clean_price(price):
  if pd.isna(price):
    return np.nan
  price_str = re.sub(r"[^\d]", "", price)
  return int(price_str) if price_str else np.nan
```

```python
df_cleaned['Price_cleaned'] = df_cleaned['Price'].apply(clean_price)
```

###### 2. Clean 'House size' column

```python
re.search(r"(\d+(?:\.\d+)?)", "250 m2").group(1)

def extract_house_size(val):
  if pd.isna(val):
    return np.nan
  match =  re.search(r"(\d+(?:\.\d+)?)", val)
  return float(match.group(1)) if match else np.nan
```

```python
df_cleaned['House_size_sqm'] = df_cleaned['House size'].apply(extract_house_size)
```

###### 3. Clean Land Size

```python
#0.5 acres

def extract_land_size(val):
  if pd.isna(val):
    return np.nan
  match = re.search(r"(\d+(?:\.\d+)?)", val)
  return float(match.group(1)) if match else np.nan
```

```python
df_cleaned['LandSize_acres'] = df_cleaned['Land size'].apply(extract_land_size)
```

##### 4. Standard and clean 'PropertyType' column

```python
#df_cleaned['propertyType'].unique()

df_cleaned['propertyType_clean'] = df_cleaned['propertyType'].str.strip().str.title()
df_cleaned['propertyType_clean'] = df_cleaned['propertyType_clean'].replace(
    {
    'Townhuse': 'Townhouse'
    }
)
```

###### 6. Standard 'Location'

```python
df_cleaned['Location'] = df_cleaned['Location'].str.strip().str.title()
```

```python
df_cleaned.head()
```

##### 8. Drop 'dirty columns' , use cleaned features

```python
cols_to_drop = ['Price','propertyType','House size','Land size']
df_cleaned.drop(columns=cols_to_drop, inplace=True)
```

```python
df_cleaned.columns = ['Location', 'Bedroom', 'bathroom', 'Price', 'House_size_sqm',
       'LandSize_acres', 'propertyType']
```
